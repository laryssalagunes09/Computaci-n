# Nivel 5

Ejercicio 20. Pregunta: ¿Cómo afecta a la puntuación del pedido el hecho de que el costo del envío sea mayor o menor al promedio?

```python
#20
df_20 = order_items.merge(order_reviews, on ="order_id", how = "left")
promedio = df_20["freight_value"].mean()

df_20["categoria_envio"] = df_20["freight_value"].apply(
    lambda x: "Envío mayor al promedio" if x > promedio else "Envío menor al promedio")

df_20.pivot_table(
    values = "review_score",
    index = "categoria_envio",
    aggfunc = ["mean", "count"]
)
```

Podemos observar que cuando el costo del envío es menor al promedio, la puntuación en las reseñas es un poco más alta (4.06) en comparación a aquillos con un envío superio al promedio (3.94).

Además, existen más pedidos realizados con un envío menor o igual al promedio.

```python
#21
df_21 = order_items.merge(products, on = "product_id", how = "left")
df_21["porcentaje_flete"] = df_21.apply(
    lambda x: (x["freight_value"] / x["price"]) * 100, axis = 1)

df_21.pivot_table(
    values = ["price", "freight_value", "porcentaje_flete"],
    index = "product_category_name",
    aggfunc= "mean"
).sort_values(by = "porcentaje_flete", ascending= False)
```

Las categorías que tienen mayor porcentaje de envío respecto al precio son productos que corresponden a artículos del hogar o multimedia, mientras que los que tienen menor porcentaje son los que tienen mayor precio, como artículos de tecnología o electrodomésticos.

## Ejercicio 22
Debido a que existían muchos nulos, simplifiqué a solo 3 categorías (entregado, no entregado y en proceso).

```python
df_22 = orders.merge(order_items, on = "order_id", how = "left")
df_22 = df_22.merge(products, on = "product_id", how = "left")

def simplificado(x):
    if x == "delivered":
        return "entregado"
    elif x in ["canceled", "unavailable"]:
        return "no entregado"
    else:
        return "en proceso"

df_22["estatus_pedido"] = df_22["order_status"].apply(simplificado)

df_22['total'] = df_22.apply(
    lambda fila: fila["price"] + fila["freight_value"],
    axis=1
)

# Cantidad de pedidos por estatus y categoría
df_22.pivot_table(
    values = "order_id",
    columns = "estatus_pedido",
    index = "product_category_name",
    aggfunc = "count"
).sort_values(by="no entregado", ascending=False)
```

```python
#22
# Costo total promedio por estatus y categoría
df_22.pivot_table(
    values = "total",
    columns = "estatus_pedido",
    index = "product_category_name",
    aggfunc = "mean"
).sort_values(by="no entregado", ascending=False)
```

```python
# Reto final
df_23 = orders.merge(order_items, on = "order_id", how = "left")
df_23 = df_23.merge(products, on = "product_id", how = "left")

def simplificado(x):
    if x == "delivered":
        return "entregado"
    elif x in ["canceled", "unavailable"]:
        return "no entregado"
    else:
        return "en proceso"

df_23["estatus_pedido"] = df_23["order_status"].apply(simplificado)

df_23['total'] = df_23.apply(
    lambda fila: fila["price"] + fila["freight_value"],
    axis=1
)

df_23["porcentaje_flete"] = df_23.apply(
    lambda x: (x["freight_value"] / x["price"]) * 100, axis = 1)

df_23.pivot_table(
    values = ["price", "freight_value", "total", "porcentaje_flete"],
    index = ["product_category_name", "estatus_pedido"],
    aggfunc= "mean"
)
```
