## Manejo de columnas
Es posible seleccionar una o varias columnas:
* Una columna
* Dos columnas
* Una lista de columnas

La sintaxis general es `df[las columnas que quiero]`

### Ejemplo para una columna

Para seleccionar la columna `col15` usamos `df["col5]`

```python
# "col5" es el nombre de la columna, no sintaxis interna de python
df[["col5"]]
```

### Ejemplo para dos columnas
Para seleccionar dos columnas la sintaxis general es `df[["col8", "col7"]]`, la diferencia con seleccionar una columna es que en esta ocasión hay doble corchete:

```python
# "col8" y "col7" son nombres de las columnas no sintaxis de python
df[["col8", "col7"]]
```

### Ejemplo para seleccionar varias columnas
En general para seleccionar varias columnas es necesario escribir la *lista* de las columnas y pasarlas con corchetes al dataframe, por ejemplo, si queremos las columnas `col1`, `col10`, `col6` y `col8`, podemos utilizar la siguiente sintaxis

```python
columnas_seleccionadas = ["col1", "col10", "col6", "col8"]
df[columnas_seleccionadas]
```

## Filtrar Datos

Al crear una condición sobre una columna estamos creando una máscara lógica (una columna de True o False por cada fila que cumpla la condición), por ejemplo `df["col6"]=="Sur"` crea una columna que será True en aquellas filas donde el valor sea "Sur" y False cuando no lo sea.

```python
df["col6"] == "Sur"
```

Para acceder a la tabla completa en la que se cumpla la condición `df["col6] == "Sur"` sólo basta con agregarlo entre corchetes al dataframe: `df[df["col6] == "Sur"]`. El resultado contiene las filas y columnas del dataframe en las cuales la columna `col16` tenga el valor "Sur"

```python
df[df["col6"] == "Sur"]
```

Para condiciones numéricas ocurre lo mismo, podemos utilizar operadores como `>, <, ==, !=, >=, <=`. Por ejemplo, para obtener las filas y columnas del dataframe que cumplan que `col1` tenga un valor mayor a 500 utilizamos `df[df["col1"] > 500]`

```python
df[df["col1"] > 500]
```

### Filtrar por más de una condición
Para filtrar por más de una condición basta con separar las condiciones por paréntesis y conectarlos con los operadores lógicos `&` (and), `|` (or), `~` (not), ejemplo:
* Las filas cuya columna 6 sea "Sur" y columna 5 diferente de "B"
    * `(df["col6"] == "Sur") & (df["col5"] != "B")`
* Las filas cuya columna 5 sea "A" o columna 6 sea "Norte"
    * `(df["col5"] == "A") | (df["col6"] == "Norte")`

```python
df[(df["col6"] == "Sur") & (df["col5"] != "B")]
```

Además, `.loc` y `.iloc` también son muy importantes y útiles para filtrar datos, `.loc` sirve para seleccionar por etiqueta y `.iloc` para seleccionar por posición. Por ejemplo, si queremos las filas de 0 a 5 y las columnas `col1` y `col2` utilizamos `df.loc[0:5, ["col1", "col2"]]`.

```python
df.loc[0:5, ["col1", "col2"]]
```

Y si queremos las filas 2 a 6 y las columnas de posición 2 a 4 utilizamos `df.iloc[2:6, 2:4]`

```python
df.iloc[2:6, 2:4]
```

## Ordenar datos

Para ordenar datos de menor a mayor utilizamos `df.sort_values()`. Si queremos que se ordene la columna 1, sería `df.sort_values("col1")`

```python
df.sort_values("col1")
```

Si queremos que ordene de mayor a menor escribimos `ascending = False`. De esta forma sería `df.sort_values("col1", ascending = False)`.
En caso de no ser numérica, ordena de A a la Z.

```python
df.sort_values("col1", ascending = False)
```

Si queremos ordenar varias columnas, las escribimos en una lista, por ejemplo para ordenar de menor a mayor primero por `col1` y después por `col2` escribimos `df.sort_values(["col1", "col2"])`.

```python
df.sort_values(["col1", "col2"])
```

## Estadísticas descriptivas

Para obtener la estadística de una columna basta con seleccionar la columna y aplicar el método, por ejemplo, la media es `.mean()`, el máximo es `.max()`, el mínimo es `.min()` la mediana `.median()` y la desviación estándar es `.std()`.

Tenemos entonces tomando de ejemplo la columna 1:
* `df["col1"].mean()` calcula la media (promedio aritmético) de la columna 1
* `df["col1"].median()` calcula la mediana de la columna 1
* `df["col1"].min()` calcula el valor más pequeño de la columna 1
* `df["col1"].max()` calcula el valor más grande de la columna 1
* `df["col1"].std()` calcula qué tan dispersos están los datos respecto a la media en la columna 1

```python
print("promedio:")
display(df["col1"].mean())

print("mediana:")
display(df["col1"].median())

print("mínimo:")
display(df["col1"].min())

print("máximo:")
display(df["col1"].max())

print("desviación estándar:")
display(df["col1"].std())
```

Además, `.describe()` proporciona estadísticas de las columnas numéricas, `count` brinda información sobre el número total de valores válidos sin contar los valores nulos. Así como promedio, desviación estándar, mínimo, máximo, primer, segundo y tercer cuartil.

```python
df.describe()
```

## Trabajar con variables categóricas

En caso de que nuestras variables sean categóricas utilizamos `value_counts()` para saber cuántas veces aparece cada categoría, por ejemplo, para la columna 6 sería `df["col6"].value_counts()`.

```python
df["col6"].value_counts()
```

Utilizamos `unique()` para saber cuáles son las categorías de una columna, para la columna 6, sería `df["col6"].unique()`

```python
df["col6"].unique()
```

Y utilizamos `nunique()` para saber cuántas categorías diferentes existen `df["col6"].nunique()`

```python
df["col6"].nunique()
```

## Agrupar datos

Podemos clasificar y agrupar por categorías en una columna y realizar operaciones para cada una de ellas con `groupby()`. Por ejemplo, si quisieramos agrupar la columa 10 y calcular el promedio de la columna 1 escribiríamos `df.groupby("col10")["col1"].mean()`. Primero escribimos la columna que queremos agrupar y luego la columna sobre la que queremos realizar la operación.

```python
df.groupby("col10")["col1"].mean()
```

También podemos calcular sobre dos o más columnas, creando una lista de columnas, en el siguiente ejemplo queremos agrupar por las categorías de la columna 10 y calcular el promedio de la columna 1 y 3 utilizando `df.groupby("col10")[["col1", "col3"]].mean()`

```python
df.groupby("col10")[["col1", "col3"]].mean()
```

Y podemos agrupar por dos o más columnas, también escribiendo una lista de columnas, por ejemplo, para clasificar por la columna 10 y 5, utilizamos `df.groupby(["col10", "col5"])[["col1", "col3"]].mean()`

```python
df.groupby(["col10", "col5"])[["col1", "col3"]].mean()
```

Para realizar conteos por categoría podemos usar `.count()`, por ejemplo para agrupar con las columnas 5, 10, y 9 y contar cuántas filas hay por cada categoría en la columna 1, podemos utilizar `df.groupby(["col5", "col10", "col9"])[["col1"]].count()`

```python
df.groupby(["col5", "col10", "col9"])[["col1"]].count()
```

## Crear y modificar columnas

Podemos crear nuevas columnas a partir de otras existentes, por ejemplo, si queremos duplicar los valores de la columna 2, utilizaremos `df["col2_duplicada"] = df["col2"] * 2`. Obtendremos una nueva columna que cumpla con las características que deseamos.

```python
df["col2_duplicada"] =df["col2"] * 2
df.head()
```

## Combinar operaciones

Después de aprender a utilizar las funciones anteriores, podemos combinarlas para crear diferentes soluciones. 
Por ejemplo, podemos agrupar por las columnas 6 y 10, y calcular el promedio de la columna 1 y tomar las que son mayores a 500.

```python
promedio = df.groupby(["col6", "col10"])[["col1"]].mean()
promedio[promedio["col1"] > 500]
```
